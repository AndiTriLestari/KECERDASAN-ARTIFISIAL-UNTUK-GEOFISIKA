| Komponen | Arti | Fungsi dalam Prompt Geofisika |
|---|---|---|
| **R** – Role | Peran yang harus dijalankan AI | Menentukan perspektif keilmuan dan tingkat analisis, misalnya sebagai geofisikawan eksplorasi, interpreter data seismik, atau analis metode geolistrik. |
| **T** – Task | Tugas spesifik yang harus dikerjakan | Menjelaskan pekerjaan analisis yang harus dilakukan, seperti menginterpretasi anomali, menghitung parameter geofisika, mengolah data, atau membandingkan hasil metode. |
| **C** – Context | Latar belakang masalah | Memberikan informasi mengenai kondisi geologi, lokasi survei, jenis metode geofisika, target eksplorasi, dan tujuan penelitian sebagai dasar interpretasi. |
| **I** – Input | Data/berkas yang tersedia untuk diproses | Menjelaskan jenis dan karakteristik data geofisika yang digunakan, seperti data resistivitas, chargeability, gravitasi, magnetik, seismik, koordinat, atau data sumur. |
| **K** – Constraints | Batasan yang tidak boleh dilanggar | Menetapkan batasan dalam pengolahan dan interpretasi geofisika, termasuk metode yang digunakan, asumsi, satuan, parameter, serta larangan membuat atau mengubah nilai data. |
| **O** – Output | Bentuk keluaran yang diminta | Menentukan bentuk hasil analisis, misalnya tabel parameter, penampang 2D, grafik, peta anomali, model bawah permukaan, atau uraian interpretasi geofisika. |
| **Q** – Quality Criteria | Kriteria agar output diterima | Menentukan standar kualitas hasil berdasarkan ketelitian perhitungan, konsistensi satuan, kesesuaian konsep geofisika, kejelasan interpretasi, dan keterkaitan dengan kondisi geologi. |
| **V** – Verification (opsional) | Cara memverifikasi hasil | Menjelaskan cara mengevaluasi keandalan interpretasi, misalnya melalui perbandingan dengan data lapangan, nilai residual, data geologi, hasil inversi, atau metode geofisika lainnya. |

## **Prompt 1 — Interpretasi Data Seismik Refleksi**

### **Role**

Bertindaklah sebagai ahli geofisika eksplorasi yang memiliki keahlian dalam metode seismik refleksi, pengolahan data seismik, analisis reflektor, serta interpretasi struktur geologi bawah permukaan.

### **Task**

Lakukan analisis data seismik refleksi untuk mengidentifikasi reflektor utama, menentukan kontinuitas lapisan, mengenali struktur geologi seperti sesar dan lipatan, serta memperkirakan kedalaman target berdasarkan informasi waktu tempuh gelombang. Analisis dilakukan melalui tahapan pemeriksaan data, identifikasi pola refleksi, analisis amplitudo, dan interpretasi penampang seismik.

### **Context**

Data seismik refleksi digunakan untuk menggambarkan kondisi geologi bawah permukaan melalui pantulan gelombang seismik pada batas lapisan yang memiliki perbedaan impedansi akustik. Interpretasi diperlukan untuk memahami geometri lapisan dan kemungkinan keberadaan struktur geologi yang dapat berhubungan dengan target eksplorasi.

### **Input**

Data yang digunakan berupa penampang seismik refleksi, informasi waktu tempuh dua arah atau Two-Way Travel Time (TWT), data kecepatan apabila tersedia, serta informasi geologi regional yang relevan. Jika data lapangan tidak tersedia, gunakan data simulasi dan nyatakan secara jelas bahwa data tersebut merupakan data simulasi.

### **Constraints**

Gunakan prinsip dasar propagasi gelombang seismik dan konsep impedansi akustik dalam interpretasi. Jangan menentukan jenis batuan hanya berdasarkan bentuk reflektor tanpa dukungan informasi tambahan. Bedakan antara reflektor yang teridentifikasi dari data, interpretasi struktur, dan asumsi yang digunakan. Gunakan istilah geofisika yang sesuai dengan tingkat mahasiswa S1 Geofisika dan jangan membuat kesimpulan yang tidak didukung oleh data.

### **Output**

Hasil analisis nantinya diarahkan pada penjelasan tahapan interpretasi, identifikasi reflektor dan struktur geologi, analisis karakteristik penampang, serta interpretasi kondisi bawah permukaan berdasarkan data seismik.

### **Quality Criteria**

Interpretasi harus konsisten dengan pola reflektor yang terlihat pada penampang, penggunaan istilah seismik harus tepat, dan setiap interpretasi struktur harus memiliki dasar yang dapat ditelusuri dari data.

### **Verification**

Periksa kembali kesesuaian interpretasi reflektor dengan pola amplitudo dan kontinuitas lapisan, konsistensi satuan waktu dan kedalaman, serta pastikan struktur yang diinterpretasikan tidak hanya didasarkan pada asumsi visual tanpa dukungan karakteristik data.

## **Prompt 2 — Analisis Data Geomagnetik**

### **Role**

Bertindaklah sebagai ahli geofisika eksplorasi yang berpengalaman dalam metode geomagnetik, pengolahan data medan magnet bumi, transformasi data anomali, dan interpretasi struktur geologi.

### **Task**

Lakukan pengolahan data geomagnetik untuk memperoleh pola anomali medan magnet yang dapat digunakan dalam identifikasi batas litologi, struktur geologi, dan kemungkinan keberadaan tubuh batuan dengan kontras suseptibilitas magnetik. Analisis mencakup pemeriksaan data awal, koreksi yang diperlukan, pemisahan anomali, transformasi data, dan interpretasi pola anomali.

### **Context**

Metode geomagnetik memanfaatkan variasi medan magnet bumi akibat perbedaan sifat kemagnetan material bawah permukaan. Variasi tersebut dapat digunakan untuk mengetahui perubahan litologi maupun keberadaan struktur geologi seperti sesar, kontak batuan, dan intrusi.

### **Input**

Masukan berupa data intensitas medan magnet total, koordinat setiap titik pengukuran, waktu pengukuran apabila tersedia, data base station jika tersedia, serta informasi geologi daerah penelitian. Data dapat diberikan dalam bentuk CSV atau Excel. Data tambahan seperti peta geologi dapat digunakan sebagai informasi pendukung.

### **Constraints**

Pertahankan satuan data asli selama tahap awal pengolahan dan jelaskan setiap perubahan satuan yang dilakukan. Gunakan koreksi yang sesuai dengan karakteristik data, termasuk koreksi variasi harian apabila data base station tersedia. Jangan menganggap anomali tinggi selalu menunjukkan batuan tertentu tanpa mempertimbangkan konteks geologi dan suseptibilitas magnetik. Pisahkan hasil pengolahan dari interpretasi.

### **Output**

Hasil analisis diarahkan pada penjelasan tahapan pengolahan data geomagnetik, karakteristik anomali medan magnet, identifikasi pola anomali yang berkaitan dengan struktur geologi, serta interpretasi kemungkinan batas litologi dan struktur bawah permukaan.

### **Quality Criteria**

Pengolahan harus mempertahankan konsistensi koordinat dan satuan, pola anomali harus sesuai dengan data masukan, dan interpretasi struktur harus memiliki hubungan yang jelas dengan karakteristik anomali magnetik.

### **Verification**

Periksa konsistensi data koordinat dan nilai medan magnet, periksa hasil koreksi dan transformasi, kemudian bandingkan pola anomali dengan informasi geologi yang tersedia sebelum membuat interpretasi akhir.

## **Prompt 3 — Analisis Data Induced Polarization (IP)**

### **Role**

Bertindaklah sebagai ahli geofisika eksplorasi mineral yang memiliki keahlian dalam metode resistivitas dan Induced Polarization (IP), pengolahan parameter chargeability, resistivitas, serta interpretasi zona mineralisasi.

### **Task**

Lakukan analisis data IP untuk mengidentifikasi distribusi chargeability dan resistivitas bawah permukaan, menentukan zona anomali, serta mengevaluasi kemungkinan keterkaitan antara anomali tersebut dengan mineralisasi bawah permukaan.

### **Context**

Metode IP digunakan untuk mengukur respons polarisasi material bawah permukaan setelah pemberian arus listrik. Variasi chargeability dan resistivitas dapat memberikan informasi mengenai kondisi material bawah permukaan, termasuk kemungkinan keberadaan mineral sulfida, lempung, atau material lain yang memiliki respons kelistrikan tertentu.

### **Input**

Data yang digunakan meliputi koordinat titik pengukuran, nilai arus, beda potensial, resistivitas semu, chargeability, konfigurasi elektroda, serta data geologi atau informasi mineralisasi apabila tersedia. Data dapat diberikan dalam bentuk tabel CSV atau Excel.

### **Constraints**

Gunakan prinsip dasar resistivitas dan Induced Polarization dalam pengolahan. Jangan menginterpretasikan nilai chargeability tinggi secara langsung sebagai zona bijih tanpa mempertimbangkan nilai resistivitas dan kondisi geologi. Pertahankan perbedaan antara data observasi, hasil pemodelan, dan interpretasi. Jika menggunakan batas anomali tertentu, jelaskan dasar penentuannya dan jangan menetapkan nilai ambang secara sembarangan.

### **Output**

Hasil analisis diarahkan pada identifikasi distribusi resistivitas dan chargeability, penentuan zona anomali, hubungan spasial antara kedua parameter tersebut, serta interpretasi zona yang memiliki indikasi mineralisasi berdasarkan dukungan data geofisika dan geologi.

### **Quality Criteria**

Hasil pengolahan harus konsisten secara matematis, satuan harus jelas, pola anomali resistivitas dan chargeability harus dapat ditelusuri dari data, serta interpretasi mineralisasi harus mempertimbangkan kemungkinan sumber anomali non-mineralisasi.

### **Verification**

Lakukan pemeriksaan terhadap satuan dan nilai parameter, konsistensi hasil perhitungan, kesesuaian pola anomali dengan distribusi titik pengukuran, serta periksa apakah interpretasi chargeability didukung oleh data resistivitas dan informasi geologi.

## **Prompt 4 — Analisis Data Ground Penetrating Radar (GPR)**

### **Role**

Bertindaklah sebagai ahli geofisika lingkungan dan geoteknik yang memiliki keahlian dalam metode Ground Penetrating Radar (GPR), pengolahan radargram, analisis reflektor, dan interpretasi struktur dangkal.

### **Task**

Lakukan pengolahan dan interpretasi data GPR untuk mengidentifikasi pola reflektor bawah permukaan, memperkirakan kedalaman objek atau batas lapisan, serta mengenali kemungkinan keberadaan utilitas, rongga, rekahan, atau perubahan material bawah permukaan.

### **Context**

GPR menggunakan gelombang elektromagnetik berfrekuensi tinggi untuk menyelidiki kondisi bawah permukaan dangkal. Perbedaan sifat elektromagnetik material dapat menghasilkan refleksi yang terlihat pada radargram. Metode ini banyak digunakan untuk investigasi dangkal karena mampu memberikan resolusi spasial yang relatif tinggi.

### **Input**

Data berupa radargram GPR, frekuensi antena, posisi lintasan, informasi kecepatan gelombang elektromagnetik apabila tersedia, serta informasi kondisi permukaan dan geologi daerah survei. Data pendukung seperti posisi objek yang telah diketahui juga dapat digunakan sebagai informasi validasi.

### **Constraints**

Lakukan interpretasi berdasarkan karakteristik reflektor, amplitudo, kontinuitas, dan pola geometrinya. Jangan langsung menyatakan suatu anomali sebagai objek tertentu tanpa dukungan karakteristik radargram. Perhatikan pengaruh kelembapan tanah, jenis material, dan frekuensi antena terhadap penetrasi dan resolusi. Bedakan antara indikasi objek dan identifikasi objek yang telah tervalidasi.

### **Output**

Hasil analisis diarahkan pada penjelasan tahapan pengolahan radargram, identifikasi reflektor dan anomali, estimasi kedalaman objek atau batas lapisan, serta interpretasi kondisi bawah permukaan dangkal berdasarkan karakteristik respons GPR.

### **Quality Criteria**

Radargram yang dianalisis harus memiliki tahapan pengolahan yang jelas, estimasi kedalaman harus menggunakan parameter kecepatan yang dapat dipertanggungjawabkan, dan interpretasi anomali harus sesuai dengan karakteristik refleksi pada data.

### **Verification**

Periksa kualitas radargram sebelum interpretasi, konsistensi parameter kecepatan dan kedalaman, hubungan antara posisi anomali dengan lintasan survei, serta lakukan validasi terhadap informasi lapangan jika tersedia.

## **Prompt 5 — Analisis Data Electrical Resistivity Tomography (ERT)**

### **Role**

Bertindaklah sebagai ahli geofisika yang memiliki keahlian dalam metode Electrical Resistivity Tomography (ERT), pengukuran resistivitas bawah permukaan, pengolahan data resistivitas, pemodelan inversi 2D, serta interpretasi kondisi geologi bawah permukaan.

### **Task**

Lakukan analisis data ERT untuk mengetahui distribusi nilai resistivitas bawah permukaan secara dua dimensi, mengidentifikasi batas antar material, serta menentukan zona yang memiliki karakteristik resistivitas berbeda. Analisis mencakup pemeriksaan data awal, pengolahan resistivitas semu, pemodelan inversi, dan interpretasi penampang resistivitas hasil inversi.

### **Context**

Metode ERT memanfaatkan injeksi arus listrik ke dalam tanah melalui elektroda dan mengukur beda potensial yang dihasilkan. Perbedaan sifat kelistrikan material bawah permukaan menghasilkan variasi nilai resistivitas yang dapat digunakan untuk membantu mengidentifikasi lapisan tanah, batuan, zona jenuh air, rekahan, atau material dengan karakteristik kelistrikan yang berbeda.

### **Input**

Data yang digunakan berupa posisi elektroda, konfigurasi elektroda, nilai arus, beda potensial, resistivitas semu, spasi elektroda, serta informasi geologi atau kondisi lapangan apabila tersedia. Data dapat diberikan dalam bentuk tabel hasil pengukuran atau file hasil survei ERT.

### **Constraints**

Gunakan prinsip dasar metode resistivitas dan konsep inversi geofisika dalam melakukan analisis. Jangan menentukan jenis material hanya berdasarkan satu nilai resistivitas karena rentang resistivitas dapat saling tumpang tindih antar material. Pertimbangkan kondisi geologi, kandungan air, porositas, dan lingkungan pengukuran dalam interpretasi. Bedakan antara nilai resistivitas hasil pengukuran, hasil inversi, dan interpretasi geologi. Jangan membuat nilai resistivitas atau parameter data yang tidak tersedia.

### **Output**

Hasil analisis nantinya diarahkan pada penjelasan tahapan pengolahan data ERT, karakteristik distribusi resistivitas bawah permukaan, hasil pemodelan inversi 2D, identifikasi zona resistivitas rendah dan tinggi, serta interpretasi kemungkinan kondisi atau material bawah permukaan berdasarkan pola resistivitas dan informasi geologi pendukung.

### **Quality Criteria**

Analisis harus menggunakan istilah ERT dan resistivitas secara tepat, memiliki hubungan yang jelas antara data pengukuran dan hasil inversi, mencantumkan satuan secara konsisten, serta membedakan hasil observasi dengan interpretasi geologi. Interpretasi harus mempertimbangkan kemungkinan adanya ambiguitas dalam hubungan antara nilai resistivitas dan jenis material.

### **Verification**

Periksa kembali konsistensi nilai dan satuan resistivitas, posisi elektroda, konfigurasi pengukuran, serta kualitas data sebelum interpretasi. Evaluasi apakah hasil inversi memberikan pola yang masuk akal dan periksa kesesuaian zona resistivitas dengan informasi geologi atau kondisi lapangan yang tersedia.
